```python
import cv2

# utils fr image visualization 
def show_image(img_path):
    img = cv2.imread(img_path)
    if img is None: 
        print("Errr: image not found")
        return 
    
    cv2.imshow("Image Window", img)   
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# Using the Nano model - not very accurate
# Here used from the command line interface yolo command 
!yolo predict model=yolo11n.pt source='https://images.pexels.com/photos/169198/pexels-photo-169198.jpeg'
```

```python
show_image('pexels-photo-169198.jpeg')
```
